```python
# 2.修改pytorch模拟线性回归中的参数或设置，比如n_samples改为500，batch_size改为32，epochs改为200，学习率改为0.001，loss_fn改为MAE，并绘制训练过程中损失值的变化曲线。

# 导入相关模块
import torch
# numpy对象 -> 张量 -> 数据集对象TensorDataset -> 数据加载器DataLoader
from torch.utils.data import TensorDataset  # 构造数据集对象
from torch.utils.data import DataLoader  # 数据加载器
from torch import nn  # nn模块中有平方损失函数和假设函数
from torch import optim  # optim模块中有优化器函数
from sklearn.datasets import make_regression  # 创建线性回归模型数据集
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei']  # 用来正常显示中文标签
plt.rcParams['axes.unicode_minus'] = False  # 用来正常显示负号

# 1.定义函数，创建线性回归数据
def create_dataset():
    # 创建数据集对象
    x, y, coef = make_regression(
        n_samples=200,          # 100条样本
        n_features=1,           # 1个特征
        noise=10,           # 噪声
        bias=13.9,            # 偏移
        coef=True,          # 是否返回系数
        random_state=6          #
    )

    # 封装成 张量
    x = torch.tensor(x, dtype=torch.float32)
    y = torch.tensor(y, dtype=torch.float32)

    # 返回结果
    return x, y, coef

# 2. 定义函数，用于模型训练
def train(x, y, coef):
    # 1. 构建DataLoader, tensor -> 数据集对象 -> 数据加载器
    dataset = TensorDataset(x, y)
    # 参数1：数据集对象，参数2：批次大小，参数3：是否打乱数据，训练集打乱，测试集不打乱，参数4：是否去掉最有一个不完整的批次
    dataloader = DataLoader(dataset, batch_size=32, shuffle=True, drop_last=False) # 100个样本，批次数16，6*16，4
    # 2. 创建模型，线性回归模型
    model = nn.Linear(1, 1)
    # 3. 创建 损失函数 和 优化器
    loss_fn = nn.MSELoss()
    optimizer = optim.SGD(model.parameters(), lr=1e-3)
    # 4.训练过程
    # 定义训练轮数
    epochs = 200
    # 记录 训练过程的 平均损失值
    total_train_loss_list = []
    # 遍历epoch
    for epoch in range(epochs):
        # 计算 每轮 的总的平均损失 和 总的样本数
        total_train_loss = 0
        total_train_sample = 0
        # 分 批次 训练
        for train_x, train_y in dataloader:
            # 前向传播，模型预测
            y_pred = model(train_x) # [batch_size, 1]
            # 计算损失值
            loss = loss_fn(y_pred, train_y.reshape(-1, 1))
            # 计算 总损失 和 总样本数
            batch_size = train_x.shape[0]
            total_train_loss += loss.item() * batch_size
            total_train_sample += batch_size
            # 梯度清零
            optimizer.zero_grad()
            # 反向传播，计算梯度
            loss.backward()
            # 更新模型参数
            optimizer.step()
        # 添加 本epoch 的平均损失 到 total_train_loss_list
        train_loss_mean = total_train_loss / total_train_sample
        total_train_loss_list.append(train_loss_mean)
        # print(f"epoch: {epoch}, 平均损失值: {train_loss_mean}")
    # 打印整个 训练损失记录
    # print(f"全部{epochs}轮的平均损失为：{total_train_loss_list}")
    print(f"模型参数：权重：{model.weight}, 偏置：{model.bias}")

    # 5. 可视化
    plt.figure(figsize=[15, 6])
    plt.subplot(1, 2, 1)
    # 绘制损失曲线
    plt.title("训练损失曲线")
    x_range = range(epochs)
    plt.plot(x_range, total_train_loss_list, label="train loss")
    # 添加网格
    plt.grid(True)
    # 添加X轴标签
    plt.xlabel("epoch")
    plt.ylabel("loss")
    # 添加曲线标签
    plt.legend()

    # 绘制预测值与真实值的对比图
    plt.subplot(1, 2, 2)
    plt.title('预测值和真实值的对比图')
    plt.scatter(x, y, label="真实数据点")
    # 计算真实的理论直线的y值
    y_true = [v*coef.item() + 13.8 for v in x]
    plt.plot(x, y_true, label="真实值")
    # 计算预测值
    # 关闭自动微分
    with torch.no_grad():
        y_pred = model(x).detach().numpy()
    # y_pred = [v*model.weight.item() + model.bias.item() for v in x]
    plt.plot(x, y_pred, label="预测值")
    # 添加网格
    plt.grid(True)
    # 添加X轴标签
    plt.xlabel("x")
    plt.ylabel("y")
    # 添加曲线标签
    plt.legend()
    plt.show()

# 测试
if __name__ == "__main__":
    x, y, coef = create_dataset()
    print(f"x.shape: {x.shape}, y.shape: {y.shape}")

    # 训练模型
    train(x, y, coef)
```

```python
# 3.绘制 prelu的激活函数图像 和 导数图像。

import torch
import matplotlib.pyplot as plt

# 1. 创建画布和坐标轴
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# -------------------------
# 第 1 个图：pReLU 函数本身
# -------------------------
x = torch.linspace(-10, 10, 1000)
a = torch.tensor([0.1])      # pReLU 的参数

y = torch.prelu(x, a)

axes[0].plot(x, y)
axes[0].set_title("pReLU 激活函数")
axes[0].grid()


# -------------------------
# 第 2 个图：pReLU 的导数
# -------------------------
x = torch.linspace(-10, 10, 1000, requires_grad=True)
a = torch.tensor([0.1])

y = torch.prelu(x, a)

# 求导
y.sum().backward()

axes[1].plot(x.detach(), x.grad)
axes[1].set_title("pReLU 激活函数导数")
axes[1].grid()

plt.tight_layout()
plt.show()
```

```python
# 2.修改ANN手机价格分类案例代码。采用周期启动的余弦退火学习率调度策略torch.optim.lr_scheduler.CosineAnnealingWarmRestarts来调整学习率，训练模型，并打印最终的测试准确率。

import torch                                                # pytorch框架
from torch.utils.data import TensorDataset                  # 数据集对象
from torch.utils.data import DataLoader                     # 数据加载器
import torch.nn as nn                                       # 神经网络层的操作
import torch.optim as optim                                 # 优化器
from sklearn.model_selection import train_test_split        # 拆分 训练集 和 测试集
import matplotlib.pyplot as plt                             # 绘图
import numpy as np                                          # 数组操作
import pandas as pd                                         # 数据处理
import time                                                 # 时间模块

from torchsummary import summary
from sklearn.preprocessing import StandardScaler

device = "cuda" if torch.cuda.is_available() else "cpu"
# 1.准备数据集
def create_dataset():
    # 1.加载CSV数据集
    data = pd.read_csv("./data/手机价格预测.csv")
    # 2.获取特征x 和 标签y
    x = data.iloc[:, :-1]
    y = data.iloc[:, -1]
    # print(f"x: {x.head()}")
    # print(f"y: {y.head()}")
    # 3.特征列 转换为 float
    x = x.astype(np.float32)
    # print(f"x: {x}")

    # 4.拆分数据集
    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=6, stratify=y)

    # 优化1：数据标准化
    transfer = StandardScaler()
    x_train = transfer.fit_transform(x_train)
    x_test = transfer.transform(x_test)

    # 5.把数据集转换为张量, numpy数组 -> 张量 -> 数据集对象TensorDataset -> 数据加载器DataLoader
    train_dataset = TensorDataset(torch.tensor(x_train), torch.tensor(y_train.values))
    test_dataset = TensorDataset(torch.tensor(x_test), torch.tensor(y_test.values))

    # 6.返回结果
    return train_dataset, test_dataset, x_train.shape[1], len(y_train.unique())

# 2.搭建神经网络
class PhonePriceClassifier(nn.Module):
    # 1.定义__init__方法, 初始化父类成员，搭建神经网络
    def __init__(self, input_dim, output_dim):
        # 初始化父类成员
        super().__init__()
        # 搭建神经网络
        # 优化2：增加网络深度
        self.linear1 = nn.Linear(input_dim, 128)
        self.linear2 = nn.Linear(128, 256)
        self.linear3 = nn.Linear(256, 512)
        self.linear4 = nn.Linear(512, 256)
        self.linear5 = nn.Linear(256, output_dim)
        pass
    # 2.定义forward方法，实现前向传播
    def forward(self, x):
        # 隐藏层1
        x = torch.relu(self.linear1(x))
        # 隐藏层2
        x = torch.relu(self.linear2(x))
        x = torch.relu(self.linear3(x))
        x = torch.relu(self.linear4(x))
        # 输出层
        x = self.linear5(x)
        # 返回结果
        return x

# 3.模型训练
def train(train_dataset, input_dim, output_dim):
    # 1.创建数据加载器DataLoader
    # 参数1：数据集对象；参数2：批量大小；参数3：是否打乱数据
    train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
    # 2.创建神经网络模型
    model = PhonePriceClassifier(input_dim, output_dim)
    model.to(device)
    # 3.定义损失函数
    loss_fn = nn.CrossEntropyLoss()
    # 4.创建优化器对象
    # 优化3：使用Adam优化方法
    optimizer = optim.Adam(model.parameters(), lr=1e-4)
    # 创建余弦退火学习率调度器
    scheduler = optim.lr_scheduler.CosineAnnealingWarmRestarts(optimizer, T_0=20, T_mult=1, eta_min=0)
    # 5.模型训练
    # 定义变量，记录 训练总轮数
    epochs = 100
    # 遍历轮数
    for epoch in range(epochs):
        # 记录 每次训练的 损失值，训练样本数
        total_loss = 0.
        total_sample = 0
        # 记录训练开始时间
        start_time = time.time()
        # 遍历批次
        for x, y in train_loader:
            # 训练模式
            model.train()
            # 移动数据到GPU
            x = x.to(device)
            y = y.to(device)
            # 前向传播
            y_pred = model(x)
            # 计算损失值
            loss = loss_fn(y_pred, y)
            # 更新总损失
            batch_size = x.shape[0]
            total_loss += loss.item()*batch_size
            # 更新总样本数
            total_sample += batch_size
            # 梯度清零
            optimizer.zero_grad()
            # 反向传播
            loss.backward()
            # 更新参数
            optimizer.step()
            # 调整学习率
            scheduler.step()

        # 打印训练信息
        # print(f"epoch: {epoch}, loss: {total_loss / total_sample}, time: {time.time() - start_time:.2f}")
        pass
    # 6.保存模型参数
    # print(f"\n\n模型的参数信息: {model.state_dict()}\n\n")
    torch.save(model.state_dict(), "./model/phone_price_classifier.pth")    # 模型文件，后缀名 .pth, .pkl, .pickle 均可
    pass

# 4.模型测试
@torch.no_grad()
def evaluate(test_dataset, input_dim, output_dim):
    # 1. 创建神经网络分类对象.
    model = PhonePriceClassifier(input_dim, output_dim)
    model.to(device)
    # 2.加载模型参数
    model.load_state_dict(torch.load('./model/phone_price_classifier.pth'))
    # 3.创建数据加载器DataLoader
    # 参数1：数据集对象；参数2：批量大小；参数3：是否打乱数据
    test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False)
    # 4.记录正确样本的数量
    correct_sample = 0
    total_sample = 0
    # 5.开始测试, 遍历数据加载器的批次数据
    for x, y in test_loader:
        # 切换到测试模式
        model.eval()
        # 张量 迁移 到 GPU
        x = x.to(device)
        y = y.to(device)
        # 6.前向传播，模型预测
        y_pred = model(x)   # 形状(batch_size, 4), 每个类别的输出logits
        # print(f"预测结果y_pred：{y_pred}")
        # 7.计算预测正确的数量
        correct = y_pred.argmax(dim=1) == y
        correct_sample += correct.sum().item()
        total_sample += y.size(0)
    print(f"测试集准确率：{correct_sample / total_sample}")

# 5.测试
if __name__ == '__main__':
    # 1.准备数据集
    train_dataset, test_dataset, input_dim, output_dim = create_dataset()
    # 打印
    print(f"train_dataset: {train_dataset}")
    print(f"test_dataset: {test_dataset}")
    print(f"input_dim: {input_dim}")
    print(f"output_dim: {output_dim}")
    # 2.构建神经网络
    model = PhonePriceClassifier(input_dim, output_dim)
    # 3.统计模型参数
    model.to(device)
    # summary(model,input_size=(input_dim,))
    # 4.模型训练
    train(train_dataset, input_dim, output_dim)
    # 5.模型测试
    evaluate(test_dataset, input_dim, output_dim)
    pass
```

```python
# 1.使用 nn.Embedding 获取128维度的词向量，输入文本为"星星
#  一颗两颗三颗四颗 连成线背著背默默许下心愿
#  看远方的星如果听的见
#  它一定实现它一定实现
#  载著你 彷彿载著阳光
#  不管到哪里都是晴天
#  蝴蝶自在飞
#  花也布满天
#  一朵一朵因你而香
#  试图让夕阳飞翔
#  带领你我环绕大自然
#  迎著风 开始共渡每一天
#  手牵手
#  一步两步三步四步望著天"，并打印。要求词表中的token按照词频从高到底排序。
import torch
import torch.nn as nn
import jieba
from collections import Counter

def demo01():
    text = '''
星星
 一颗两颗三颗四颗 连成线背著背默默许下心愿
 看远方的星如果听的见
 它一定实现它一定实现
 载著你 彷彿载著阳光
 不管到哪里都是晴天
 蝴蝶自在飞
 花也布满天
 一朵一朵因你而香
 试图让夕阳飞翔
 带领你我环绕大自然
 迎著风 开始共渡每一天
 手牵手
 一步两步三步四步望著天
'''

    # 2.使用jieba 进行分词
    words = jieba.lcut(text)
    print(f"词序列words: {words}")

    # === 按词频排序 ===
    word_freq = Counter(words)
    unique_words = [w for w, _ in word_freq.most_common()]
    print(f"按照词频排序后的unique_words: {unique_words}")

    # === 构建词表 ===
    vocab = {word: index for index, word in enumerate(unique_words)}
    print(f"vocab: {vocab}")

    # 4.创建词嵌入层
    embed = nn.Embedding(num_embeddings=len(vocab), embedding_dim=128)
    print(f"embed: {embed}")

    # 5.将每个词转换为词向量
    for word in words:
        idx = vocab[word]
        idx_tensor = torch.tensor([idx])
        word_vector = embed(idx_tensor)
        print(f"{idx}: {word}\t词向量: {word_vector.detach()}")

if __name__ == '__main__':
    demo01()
```
